```python
import pandas as pd
import numpy as np
import datetime

np.random.seed(42)  # makes the random numbers the same every time you run it

# --- Config ---
MIN_DETECTIONS            = 20
ONSET_THRESHOLD           = 0.05
END_THRESHOLD             = 0.95
MIN_SEASON_FOR_BIMODALITY = 90
TROUGH_THRESHOLD          = 0.30
BC_THRESHOLD              = 0.555

df_full = pd.read_csv("outputs/med_basin_v2/daily_counts/745_Saharan_halophytics_daily.csv")
print(df_full.groupby('year')['n_detections'].sum())
```

```python
# Good year — plenty of detections
df_2003 = df_full[df_full['year'] == 2003][['doy', 'n_detections']].sort_values('doy').reset_index(drop=True)

# Sparse year — below MIN_DETECTIONS (15 detections, threshold is 20)
df_2008 = df_full[df_full['year'] == 2008][['doy', 'n_detections']].sort_values('doy').reset_index(drop=True)
```
